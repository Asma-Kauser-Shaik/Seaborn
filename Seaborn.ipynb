```python
# How to load the dataset from the seaborn
df = sns.load_dataset('flights')
df
```

# Line chart in seaborn

```python
# USE: trends over the time
# group the fight dataset yearwise
```

```python
# groupby function() to group the yearwise data.
timeline_year = df.groupby("year")["passengers"].sum().reset_index()
timeline_year
```

```python
# line chart using seaborn
timeline_year = df.groupby("year")["passengers"].sum().reset_index()
sns.lineplot(x='year',y='passengers',data=timeline_year,marker = "o")
```

```python
# you can customize the seaborn chart using matplotlib
import matplotlib.pyplot as plt
timeline_year = df.groupby("year")["passengers"].sum().reset_index()
sns.lineplot(x='year',y='passengers',data=timeline_year,marker = "o")
plt.grid()
```

# Pairplot

```python
# pairplot : par plot shows the relationship between all the numerical variable together pair wise
# this plot is which can ploted directly in seaborn
# not in matplotlib
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue ='sex')# hue parameter  is used for categorize the data.
```

```python
sns.pairplot(df,hue='day')
```

# Heatmap

```python
# Heapmap is used to visulaize the corrleation between 2 variables in colorfull chart.
# df.corr()works on numericals data
corr_matrix = df.corr(numeric_only = True)
print(corr_matrix)
```

```python
plt.figure(figsize=(3,2))
sns.heatmap(corr_matrix,annot = True,cmap ="coolwarm")
```

# Relationship between categorical columns

```python
# using pivot table : to find relationship between categorical variables
# heapmap : make the colorful for showing pivot table
# categorical columns : sex,day,time,smoker
```

```python
pivot = df.pivot_table(index = "day",columns = "sex",aggfunc = "size")
pivot
```

```python
plt.figure(figsize = (3,2))
sns.heatmap(pivot,annot = True,cmap = "Reds")
```

```python
pivot = df.pivot_table(index =[ "day","time"],columns = ["sex","smoker"],aggfunc = "size")
pivot
```

```python
sns.heatmap(pivot,annot = True,cmap = "Greens")
```

```python
#countplot : showing frequency of cateogrical data.
sns.countplot(data = df,x = "day",hue = "sex")
```
